# Semana 1 · Extracción de fotogramas

Se extraen los diez vídeos ZIP en cinco lotes, a resolución original. La misma
implementación se encuentra en `herramientas/preparar_fotogramas.py`: recorre los
fotogramas desde 1 y guarda cada múltiplo de `INTERVAL`. No genera XML: anotar una
imagen y marcarla como negativo son decisiones de quien la revisa.

Ejecutar dentro del repositorio. Se requiere OpenCV. Si se ejecuta en JupyterHub,
clonar el repositorio o descomprimir su ZIP; la carpeta de trabajo puede ser WEEK1.
El empaquetador permite importar después las secuencias a `/workspace/data_train`.


```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
RAIZ = next((p for p in (cwd, *cwd.parents)
             if (p / 'herramientas/preparar_fotogramas.py').is_file()), None)
if RAIZ is None:
    raise FileNotFoundError('Ejecuta el notebook dentro del repositorio PIDS-Proyecto2')
sys.path.insert(0, str(RAIZ / 'herramientas'))
from preparar_fotogramas import preparar, comprobar
```

```python
ORIGEN = RAIZ / 'ENTREGA 1/WEEK1/DATASET PIDS'
DESTINO = RAIZ / 'ENTREGA 1/WEEK1/ANOTACION'
INTERVAL = 24
```

```python
if DESTINO.exists():
    resumenes, registros = comprobar(DESTINO, ORIGEN)
    if any(v['intervalo'] != INTERVAL for v in resumenes):
        raise ValueError('El dataset existente usa otro intervalo; elige un DESTINO nuevo')
    print('Dataset existente verificado; se conservan imágenes y anotaciones.')
else:
    resumenes, registros = preparar(ORIGEN, DESTINO, INTERVAL)
    comprobar(DESTINO, ORIGEN)
```

```python
print(f"{'Vídeo':<24} {'Frames':>7} {'FPS':>8} {'Duración':>10} {'Imágenes':>9}")
for v in resumenes:
    print(f"{v['video']:<24} {v['fotogramas']:>7} {v['fps']:>8.3f} {v['duracion_s']:>9.2f}s {v['imagenes']:>9}")
print(f'Total: {len(registros)} imágenes originales')
```
